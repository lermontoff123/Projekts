## Анализ данных роботизированных звонков и предложения по оптимизации

#### 1. Анализ контактности по часам

```python
# Фильтруем только успешные контакты (Intention A) и считаем по часам
success_calls = df[df['Intention'] == 'A']
contact_hours = success_calls['Call_Hour'].value_counts().sort_index()
total_hours = df['Call_Hour'].value_counts().sort_index()
contact_rate = (contact_hours / total_hours * 100).fillna(0).round(1)  # Округляем до 1 знака

plt.figure(figsize=(14, 7))
# Исправленная строка с barplot - добавлен hue и legend=False
ax = sns.barplot(x=contact_rate.index, 
                 y=contact_rate.values, 
                 hue=contact_rate.index,  # Добавляем hue
                 palette="viridis",
                 legend=False)           # Отключаем легенду

# Вычисляем средний показатель
mean_rate = contact_rate.mean()

# Аннотации и оформление
plt.title('Эффективность контактов по часам дня\n(Процент успешных обещаний оплаты)', fontsize=16, pad=20)
plt.xlabel('Час дня', fontsize=12, labelpad=10)
plt.ylabel('Успешность (%)', fontsize=12, labelpad=10)
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.3)  # Добавляем сетку

# Линия среднего и её подпись
plt.axhline(mean_rate, color='red', linestyle='--', linewidth=1, alpha=0.7)
ax.text(len(contact_rate)-0.5, mean_rate+1, f'Среднее: {mean_rate:.1f}%', 
        color='red', ha='right', va='bottom', fontsize=10)

# Добавляем подписи значений на столбцы
for p in ax.patches:
    # Динамическое позиционирование текста
    offset = 1 if p.get_height() < 5 else -10 if p.get_height() > 90 else 2
    va = 'bottom' if p.get_height() < 5 else 'top'
    
    ax.annotate(f"{p.get_height():.1f}%", 
               (p.get_x() + p.get_width() / 2., p.get_height()),
               ha='center', 
               va=va,
               xytext=(0, offset), 
               textcoords='offset points',
               fontsize=10,
               bbox=dict(boxstyle="round,pad=0.3", 
                         fc="white",
                         ec="none",
                         lw=0,
                         alpha=0.8))

plt.tight_layout()
plt.show()
```

##### Вывод
##### Пик эффективности наблюдается в утренние часы (7-8) и около 16:00. Рекомендуется увеличить долю звонков в эти периоды.

#### 3. Влияние времени ответа на успешность

```python
# Группируем по времени ответа (секунды)
bins = [0, 5, 10, 15, 20, 30, 60, 120]
labels = ['0-5', '5-10', '10-15', '15-20', '20-30', '30-60', '60+']
df['Answer_Delay_Group'] = pd.cut(df['Answer_Delay'], bins=bins, labels=labels)

success_by_delay = df[df['Intention'] == 'A'].groupby('Answer_Delay_Group', observed=True).size()
total_by_delay = df.groupby('Answer_Delay_Group', observed=True).size()
rate_by_delay = (success_by_delay / total_by_delay * 100).fillna(0).round(1)

# Создаем график с увеличенной осью Y
plt.figure(figsize=(14, 8))
ax = sns.lineplot(x=rate_by_delay.index, y=rate_by_delay.values, 
                 marker='o', markersize=10, linewidth=3, color='darkorange')

# Увеличиваем пределы оси Y (добавляем 10% запаса сверху)
ymax = rate_by_delay.max() * 1.1
plt.ylim(0, ymax*1.1)

# Оформление графика
plt.title('Зависимость успешности контактов от времени ответа клиента', 
          fontsize=16, pad=20)
plt.xlabel('Время ответа (секунды)', fontsize=13, labelpad=12)
plt.ylabel('Процент успешных контактов (%)', fontsize=13, labelpad=12)
plt.grid(True, alpha=0.3)

# Добавляем подписи значений
for x, y in zip(rate_by_delay.index, rate_by_delay.values):
    ax.text(x, y+1.5, f'{y:.1f}%', 
            ha='center', va='bottom', fontsize=11,
            bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="none", alpha=0.8))

# Аннотируем пиковые значения
max_idx = rate_by_delay.idxmax()
max_val = rate_by_delay.max()
plt.annotate(f'Пик: {max_val:.1f}%', 
             xy=(max_idx, max_val), 
             xytext=(15, 15),
             textcoords='offset points', 
             ha='center',
             fontsize=12,
             bbox=dict(boxstyle="round,pad=0.5", fc="yellow", alpha=0.5),
             arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=0.2"))

# Добавляем и настраиваем линию среднего
mean_rate = rate_by_delay.mean()
plt.axhline(mean_rate, color='red', linestyle='--', alpha=0.7, linewidth=2)
# Смещаем подпись среднего левее и ниже
ax.text(len(rate_by_delay)-1.5, mean_rate+0.5, f'Среднее: {mean_rate:.1f}%',  # x=0.5 - смещение к началу графика
        color='red', ha='left', va='top', fontsize=11,
        bbox=dict(boxstyle="round,pad=0.3", fc="white", alpha=0.8))

plt.tight_layout()
plt.show()
```

#### 8. Эффективность разных типов роботов

##### Гипотеза:
Версии роботов с дополнительными модификациями (PTP) показывают лучшую результативность, возможно, благодаря усовершенствованным сценариям.

```python
# Создаем DataFrame с данными по каждому роботу
robot_stats = pd.DataFrame({
    'total_calls': df['Robot'].value_counts(),
    'success_calls': df[df['Intention'] == 'A']['Robot'].value_counts()
}).fillna(0)

# Рассчитываем процент успешных контактов с доверительным интервалом
robot_stats['success_rate'] = (robot_stats['success_calls'] / robot_stats['total_calls'] * 100)
robot_stats = robot_stats.sort_values('success_rate', ascending=False)

# Рассчитываем доверительные интервалы (Wilson score interval)
import statsmodels.stats.proportion as sm
ci_low = []
ci_high = []

for idx, row in robot_stats.iterrows():
    ci = sm.proportion_confint(row['success_calls'], row['total_calls'], alpha=0.05, method='wilson')
    ci_low.append(ci[0]*100)
    ci_high.append(ci[1]*100)

robot_stats['ci_low'] = ci_low
robot_stats['ci_high'] = ci_high

# Визуализация с учетом объема данных
plt.figure(figsize=(12, 7))
ax = plt.gca()

# Цвета в зависимости от объема звонков
colors = ['#1f77b4' if x < robot_stats['total_calls'].quantile(0.5) 
          else '#ff7f0e' if x < robot_stats['total_calls'].quantile(0.75)
          else '#2ca02c' for x in robot_stats['total_calls']]

# График с доверительными интервалами
bars = ax.bar(robot_stats.index, robot_stats['success_rate'], 
              color=colors, alpha=0.7,
              yerr=[robot_stats['success_rate'] - robot_stats['ci_low'], 
                    robot_stats['ci_high'] - robot_stats['success_rate']],
              error_kw={'elinewidth': 2, 'capthick': 2, 'capsize': 5})

# Оформление
plt.title('Эффективность разных версий роботов\n(с учетом объема выборки и доверительных интервалов)', fontsize=16)
plt.xlabel('Тип робота', fontsize=12)
plt.ylabel('Процент успешных контактов (%)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

# Линия среднего
mean_rate = (robot_stats['success_calls'].sum() / robot_stats['total_calls'].sum()) * 100
ax.axhline(mean_rate, color='red', linestyle='--', linewidth=1.5, label=f'Общее среднее: {mean_rate:.1f}%')

# Аннотации с объемом данных
for i, bar in enumerate(bars):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + 1,
            f"{robot_stats.iloc[i]['total_calls']:,}\nзвонков",
            ha='center', va='bottom', fontsize=9)

# Легенда для объема данных
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#1f77b4', label='Малый объем (<50%)'),
    Patch(facecolor='#ff7f0e', label='Средний объем (50-75%)'),
    Patch(facecolor='#2ca02c', label='Большой объем (>75%)')
]
ax.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# Вывод таблицы с полной статистикой
display(robot_stats[['total_calls', 'success_calls', 'success_rate', 'ci_low', 'ci_high']]
        .sort_values('success_rate', ascending=False)
        .style.format({
            'total_calls': '{:,}',
            'success_calls': '{:,}',
            'success_rate': '{:.1f}%',
            'ci_low': '{:.1f}%',
            'ci_high': '{:.1f}%'
        }).background_gradient(cmap='Blues', subset=['success_rate']))
```

##### Вывод:
##### Модификация PTP показывает в 7 раз лучшие результаты (возможно, из-за малого количества обработанных вызовов).
##### Рекомендуется проанализировать её сценарий для масштабирования успешных практик.

#### 11. Зависимость от времени до платежа (DueDate)

Клиенты ближе к дате платежа более склонны к обещанию оплаты.

```python
# Вычисляем дни до платежа с обработкой ошибок
# Пробуем преобразовать даты с указанием формата (укажите ваш реальный формат дат)
df['days_to_due'] = (pd.to_datetime(df['DueDate/due_date'], format='%Y-%m-%d', errors='coerce') - 
                     pd.to_datetime(df['Call_Day'], format='%Y-%m-%d', errors='coerce')).dt.days
    
# Удаляем строки с некорректными датами
df1 = df.dropna(subset=['days_to_due'])
    
# Группируем по дням до платежа
success_days = df1[df1['Intention'] == 'A'].groupby('days_to_due').size()
total_days = df1.groupby('days_to_due').size()
days_rate = (success_days / total_days * 100).fillna(0)

plt.figure(figsize=(14, 6))
sns.regplot(x=days_rate.index, y=days_rate.values, 
                scatter_kws={'alpha':0.5, 's':80}, 
                line_kws={'color':'red', 'linewidth':2}, 
                lowess=True)

# Оформление
plt.title('Эффективность контактов в зависимости от дней до платежа', fontsize=16)
plt.xlabel('Дней до даты платежа', fontsize=12)
plt.ylabel('Процент успешных контактов (%)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xlim(-5, 6)
plt.ylim(0, days_rate.max()*1.1)  # Добавляем 10% отступа сверху

plt.tight_layout()
plt.show()
```

##### Вывод:
Максимальная эффективность через 4-5 дней после просрочки платежа. Контакты раньше дня оплаты не так эффективны.
